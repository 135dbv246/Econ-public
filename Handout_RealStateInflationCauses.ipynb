### **Econ 101 Vocabulary & Key Concepts**
* **Price Elasticity of Supply ($E_s$):** A measure of how responsive the quantity supplied of a good (housing) is to a change in its price. When $E_s < 1$, supply is *inelastic*.
* **Approval Lags & Frictions:** Administrative delays, public hearings, and procedural hurdles that lengthen project timelines, increasing carrying costs and uncertainty for suppliers.
* **Regulatory Tax:** The implicit price wedge created by government land-use controls that pushes final home prices significantly above physical construction and land acquisition costs.
* **Regulatory Arbitrage:** Restructuring a business project or development design to take advantage of favorable state or legal provisions that bypass restrictive local regulations.
* **Second-Order / Unintended Consequences:** Unforeseen secondary outcomes resulting from a policy intervention (e.g., price controls creating municipal tax shifts).

## **1. Developer Strategy Case Study: The South LA Costco Mixed-Use Project**
In high-cost, highly regulated urban markets such as California, residential real estate development is frequently stymied by administrative delays, public hearings, and local land-use constraints[cite: 9]. A prominent real-world example of navigating these hurdles is the proposed mixed-use Costco development in South Los Angeles (Baldwin Village), led by developer Thrive Living[cite: 9].

Developing standalone commercial big-box retail space in California routinely triggers discretionary reviews and extensive environmental impact reports under the California Environmental Quality Act (CEQA)[cite: 9]. To bypass these lengthy approval bottlenecks, the developer utilized state-level legislation—specifically **Assembly Bill 2011 (AB 2011)**[cite: 9].

By designing a mixed-use complex that incorporates 800 residential apartment units above the ground-floor warehouse store (ensuring residential space comprises at least two-thirds of the total square footage), the project qualified for "by-right" streamlined ministerial approval[cite: 9]. This regulatory arbitrage strategy allowed the project to circumvent discretionary local planning boards and public review delays, transforming an otherwise uncertain multi-year permitting process into an expedited development path[cite: 9].

```python
# @title Developer Strategy Simulator - DO: Change % resident share
import ipywidgets as widgets
from IPython.display import display, HTML

def simulate_ab2011(residential_share):
    qualifies = residential_share >= 66.7
    months = 6 if qualifies else 36
    carrying_cost = months * 150000  # Estimated monthly carrying cost / interest

    status_color = "green" if qualifies else "red"
    status_text = "QUALIFIES for AB 2011 By-Right Approval" if qualifies else "DOES NOT QUALIFY (Subject to discretionary review & CEQA)"

    html = f"""
    <div style="border: 2px solid {status_color}; padding: 15px; border-radius: 8px; background-color: #f9f9f9;">
        <h4 style="color: {status_color}; margin-top:0;">Project Status: {status_text}</h4>
        <ul>
            <li><b>Residential Share:</b> {residential_share:.1f}%</li>
            <li><b>Estimated Approval Lag:</b> <span style="font-size: 16px; font-weight: bold;">{months} Months</span></li>
            <li><b>Estimated Developer Holding Cost:</b> <span style="font-size: 16px; font-weight: bold; color: {status_color};">${carrying_cost:,.0f}</span></li>
        </ul>
        <p><i>Notice how meeting the 2/3 (66.7%) residential threshold eliminates discretionary public hearings, drastically lowering project holding costs.</i></p>
    </div>
    """
    display(HTML(html))

print("=== AB 2011 Developer Strategy Simulator ===")
widgets.interact(
    simulate_ab2011,
    residential_share=widgets.FloatSlider(
        value=50.0,
        min=10.0,
        max=90.0,
        step=1.0,
        description='Res. Share %:',
        style={'description_width': 'initial'}
    )
);
```

## **2. Policy & Local Market Dynamics: Saint Paul’s Rent Control vs. Missing Middle Upzoning**
Local housing policy often involves competing tools that yield conflicting market signals[cite: 8]. Analysis by the Federal Reserve Bank of Minneapolis (2026) highlights this dynamic in Saint Paul, Minnesota, which simultaneously implemented zoning expansion and strict price controls[cite: 8].

### **Upzoning for "Missing Middle" Housing**
In 2023, Saint Paul amended its zoning code to establish H1 and H2 residential districts[cite: 8]. This reform legalized "missing middle" housing (duplexes, triplexes, fourplexes, and up to sixplexes with affordability incentives) on residential lots citywide[cite: 8]. By expanding allowable land density by-right, the city permitted 76 missing-middle units in 2024 alone—matching the total number of such units permitted over the entire preceding decade[cite: 8].

### **Rent Control Price Ceilings & Second-Order Effects**
In November 2021, Saint Paul voters approved a 3% annual cap on residential rent increases[cite: 8]. The empirical results demonstrate major second-order consequences:
* **Multifamily Permitting Collapse:** Annual housing permits plunged from peak years of over 2,000 units in 2020–2021 down to 404 in 2024 and 357 in 2025 as private capital flows and tax-credit syndicators froze investments[cite: 8].
* **Development Bottlenecks:** High-profile projects, such as market-rate housing at the 135-acre Highland Bridge redevelopment site, stalled completely until city leaders passed amendments permanently exempting new construction[cite: 8].
* **Tax Base Shift to Homeowners:** Between 2020 and 2025, median per-unit apartment sale prices fell 36%, driving total aggregate apartment property taxes down 27% (from $141 million to $103 million)[cite: 8]. To cover local government budgets, the tax burden shifted onto single-family homeowners, whose annual property taxes rose from $3,442 to $4,264[cite: 8].

```python
# @title Shifters Impact on Prices (Es>1 or Es<1 cases)
import matplotlib.pyplot as plt
import numpy as np
import ipywidgets as widgets

def plot_housing_market(demand_shift=0, supply_shift=0, elasticity="Inelastic"):
    # Define quantity vector
    Q = np.linspace(0, 100, 500)

    # Inelastic supply has a steeper slope; Elastic supply has a flatter slope
    slope = 1.5 if elasticity == "Inelastic" else 0.4

    # Base Curves
    P_demand = 110 - Q + demand_shift
    P_supply = 15 + slope * Q + supply_shift

    # Create plot figure
    plt.figure(figsize=(9, 5.5))
    plt.plot(Q, P_demand, label="Demand (D)", color="#1f77b4", linewidth=2.5)
    plt.plot(Q, P_supply, label="Supply (S)", color="#d62728", linewidth=2.5)

    # Calculate Equilibrium
    Q_eq = (95 + demand_shift - supply_shift) / (1 + slope)
    P_eq = 110 - Q_eq + demand_shift

    # Plot Equilibrium Point
    if 0 <= Q_eq <= 100 and 0 <= P_eq <= 150:
        plt.scatter([Q_eq], [P_eq], color="black", zorder=5, s=70)
        plt.annotate(f"Equilibrium\n(P=${P_eq:.1f}, Q={Q_eq:.1f})",
                     (Q_eq + 2, P_eq + 2), fontsize=10, fontweight='bold')

    plt.title(f"Metropolitan Housing Market Simulator ({elasticity} Supply)", fontsize=13, fontweight='bold')
    plt.xlabel("Quantity of Housing Units (Q)", fontsize=11)
    plt.ylabel("Rent / House Price Level (P)", fontsize=11)
    plt.ylim(0, 160)
    plt.xlim(0, 100)
    plt.grid(True, linestyle="--", alpha=0.5)
    plt.legend(fontsize=11)
    plt.show()

print("=== Interactive Market Equilibrium Simulator ===")
widgets.interact(
    plot_housing_market,
    demand_shift=widgets.IntSlider(min=0, max=40, step=5, value=0, description="Demand Shift (+D):"),
    supply_shift=widgets.IntSlider(min=0, max=40, step=5, value=0, description="Zoning Friction (-S):"),
    elasticity=widgets.Dropdown(options=["Inelastic", "Elastic"], value="Inelastic", description="Supply Type:")
);
```

## **3. Academic Literature Context: Gyourko & Molloy (2014)**
The microeconomic mechanisms underlying these policy outcomes directly align with academic literature evaluating residential land-use regulations (Gyourko & Molloy, 2014)[cite: 2, 4, 9].

### **Key Academic Principles**
* **Approval Lags & Uncertainty:** Regulations create administrative delays ("approval lags") and political uncertainty, inflating developer risks and carrying costs while choking supply responses[cite: 2, 4, 9].
* **The "Regulatory Tax":** Strict land-use controls act as an implicit tax, creating a substantial price wedge between physical building costs and market sales prices[cite: 2, 4, 9].
* **Inelasticity of Supply:** Regulatory hurdles lower the price elasticity of housing supply[cite: 2, 4, 9]. When market demand increases, rigid supply forces adjustment through price inflation rather than unit expansion[cite: 2, 4, 9].

## **4. Macroeconomic Trends: St. Louis Fed Analysis (García & Garriga, 2026)**
Analyzing macroeconomic data across nearly 1,500 U.S. counties, Federal Reserve Bank of St. Louis economists Manu García and Carlos Garriga (2026) document the decoupling of home prices from household paychecks[cite: 3, 4, 9].

> *"The core economic mechanism is straightforward: Demand shifts outward against a relatively inelastic supply... construction delays and zoning restrictions push demand primarily into higher prices rather than more housing units."*[cite: 3, 4, 9]

### **Analysis of Figure 1 (House-Price-to-Income Scatter Plot)**
* **Understanding the 45-Degree Line:** The 45-degree line represents equal price-to-income ratios between 2000 and 2023 ($\text{Ratio}_{2023} = \text{Ratio}_{2000}$)[cite: 3, 4, 5, 9]. Points lying above the line indicate worsening affordability relative to earnings[cite: 3, 4, 5, 9].
* **Key Takeaway:** The cloud of county data points sits almost entirely **above the 45-degree line**, proving that deteriorating housing affordability is a systemic nationwide phenomenon driven by inelastic supply[cite: 3, 4, 5, 9].

## **5. Comparative Policy Analysis & Natural Experiments (Mises Institute, 2020)**
Synthesizing research on regulatory policy, Harvey & Alex (2020) examine cross-jurisdictional "natural experiments" that isolate policy variables from macro demand drivers[cite: 7, 9].

### **Debunking Demand-Side Scapegoats**
Critics frequently attribute housing price surges to population growth, income growth, or inflation[cite: 7, 9]. Empirical comparisons refute these claims:
* **Population Growth:** In the 1970s, Palo Alto home prices quadrupled while population growth was near zero[cite: 7, 9]. Conversely, Las Vegas's population tripled between 1980 and 2000 with zero real median price increase due to flexible supply expansion[cite: 7, 9].
* **Income Growth:** Dallas and Houston sustain household income growth 10% above the national average while maintaining home prices below the national average[cite: 7, 9].

### **California vs. Texas Natural Experiment**
Comparing America's two largest state economies reveals the structural impact of land-use frameworks[cite: 7, 9]. Over the past two decades, Texas experienced faster population growth and identical real GDP growth compared to California[cite: 7, 9]. However, California home prices escalated significantly faster[cite: 7, 9]. Texas benefits from flexible municipal land policy (including Houston's unzoned framework, ranked 24th worldwide for affordability), while California features 17 of America's least affordable markets due to extensive discretionary reviews and zoning restrictions[cite: 7, 9].

### **International Zoning Frameworks: Tokyo vs. London**
Evaluating global mega-cities demonstrates the impact of dynamic zoning[cite: 7, 9]. Despite a population density of over 6,100 people per $\text{km}^2$, Tokyo ranks 16th worldwide in house price levels, whereas London ranks 5th[cite: 7, 9]. Japan utilizes national, nuisance-based zoning that permits mixed-use construction by-right across almost all zones and allows malleable height limits[cite: 7, 9]. Consequently, Tokyo recorded 942,000 housing starts in 2018 (compared to 194,000 in the UK), maintaining flat real property prices over two decades[cite: 7, 9].

  * *Shifts:* Demand shifts rightward ($D_1 \rightarrow D_2$) due to population growth[cite: 9]. Supply shifts leftward ($S_1 \rightarrow S_2$) due to zoning restrictions and regulatory costs[cite: 9].  
  * *Outcomes:* **Equilibrium Price ($P$) Definitively Increases ($P_2 > P_1$)** because both shifts push prices up[cite: 9]. **Equilibrium Quantity ($Q$) is Indeterminate / Ambiguous** because demand increases $Q$ while supply decreases $Q$ (net change depends on relative shift magnitudes)[cite: 9].
